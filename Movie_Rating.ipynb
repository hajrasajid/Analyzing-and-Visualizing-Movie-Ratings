```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['IMDB_Rating'], bins=10)

plt.title("Distribution of IMDB Ratings")
plt.xlabel("Ratings")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.boxplot(x=df['IMDB_Rating'])

plt.title("Boxplot of Ratings")

plt.show()
```

```python
top_movies = df.sort_values(by='IMDB_Rating', ascending=False)

top_movies[['Series_Title', 'IMDB_Rating']].head(10)
```

```python
genre_rating = df.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False)

print(genre_rating.head(10))
```

```python
import matplotlib.pyplot as plt

genre_rating.head(10).plot(kind='bar', figsize=(10,5))

plt.title("Top Rated Genres (Average IMDB Rating)")
plt.xlabel("Genre")
plt.ylabel("Average Rating")

plt.show()
```
